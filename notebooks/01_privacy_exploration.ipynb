# VeilSight: privacy exploration

This notebook is for inspecting the pipeline, not for running the product. Images render inline so no native OpenCV window can hang the kernel.

```python
from pathlib import Path
import sys

import cv2
import matplotlib.pyplot as plt

sys.path.insert(0, str(Path.cwd().parent / 'backend'))
from app.detectors.haar import HaarFaceDetector
from app.anonymization import anonymize_faces
```

```python
image_path = Path.cwd().parent / 'frontend' / 'public' / 'sample-lab.png'
image = cv2.imread(str(image_path))
assert image is not None, image_path

faces = HaarFaceDetector().detect(image, confidence=0.55)
protected = anonymize_faces(image, faces, method='blur', intensity=0.82)
print(f'Detected {len(faces)} faces')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
for axis, frame, title in zip(axes, [image, protected], ['Original', 'Protected']):
    axis.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    axis.set_title(title)
    axis.axis('off')
plt.tight_layout()
plt.show()
```
